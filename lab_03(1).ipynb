## Часть 2. Распределение студентов по курсам

Теперь вам нужно распределить студентов по осенним курсам по выбору, учитывая их предпочтения.

Алгоритм распределения студентов по курсам:
1. По умолчанию на каждой дисциплине по выбору у 2 и 3 курсов может учиться 1 группа (до 30 студентов). Исключения описаны ниже. На blended-дисциплинах для третьекурсников количество мест не ограничено.
2. Проводится первая волна отбора. Для каждой дисциплины формируется список тех, кто указал её первым приоритетом (если студент должен выбрать два курса по выбору, то для него дисциплины, которые он указал первым и вторым приоритетом, рассматриваются как дисциплины первого приоритета). Если желающих больше, чем мест, то выбирается топ по перцентилю рейтинга.
3. На дисциплинах, где остались места после первой волны, формируются списки тех, кто выбрал их вторым приоритетом, и места заполняются лучшими по перцентили рейтинга студентами. После этого проводится такая же процедура для дисциплин третьего приоритета.
4. Если студент не попал на необходимое количество курсов по итогам трёх волн, с ним связывается учебный отдел и решает вопрос в индивидуальном порядке.

Обращаем ваше внимание на следующие детали:

- По умолчанию студент выбирает один осенний и один весенний курс по выбору, а также третьекурсники выбирают один blended-курс. Студенты групп третьего курса специализаций МОП и ТИ выбирают по 2 осенних и 2 весенних курса по выбору, также студенты групп второго курса специализации МИ выбирают 2 осенних курса. <i>Для студентов, которые выбирают 2 курса (например, осенних) первый приоритет — <code>fall_1</code> и <code>fall_2</code>, второй приоритет — <code>fall_3</code>. Такие студенты участвуют только в двух волнах отбора</i>.

- Студенты специализации МОП не могут выбрать весенним курсом по выбору Машинное обучение 2. <i>Если студент специализации МОП выбрал Машинное обучение 2, то его приоритеты сдвигаются. Из-за совпадений первого и второго курса по выбору двигать приоритеты не надо</i>.

- Blended-курсы не трогайте, по ним не надо распределять, на другие курсы они никак не влияют.

Если вы придумаете, как обойтись и без этих циклов, то на усмотрение проверяющего могут быть добавлены бонусные баллы. <b>Дублирование кода не признается успешным избавлением от циклов</b></i>

Проведите все три волны отбора студентов на курсы по выбору
Результат сохраните в файл res_fall.csv.

```python

# Заявки студентов, которым нужен один осенний курс
first_priority_one = df.loc[
    df['fall_course_num'] == 1,
    ['id', 'percentile', 'fall_1']
].rename(columns={'fall_1': 'course'})

# Заявки студентов, которым нужны два осенних курса
first_priority_two = (
    df.loc[
        df['fall_course_num'] == 2,
        ['id', 'percentile', 'fall_1', 'fall_2']
    ]
    .melt(
        id_vars=['id', 'percentile'],
        value_name='course'
    )
    [['id', 'percentile', 'course']]
)

# Объединяем все заявки первой волны
first_priority = pd.concat(
    [first_priority_one, first_priority_two],
    ignore_index=True
)

# Убираем пустые курсы и возможные повторные заявки
first_priority = (
    first_priority
    .dropna(subset=['course'])
    .query("course != ''")
    .drop_duplicates(subset=['id', 'course'])
)
```

```python

course_capacity = {
    'Statistical Learning Theory': 60,
    'Высокопроизводительные вычисления': 60,
    'Анализ неструктурированных данных': np.inf
}

first_priority['capacity'] = (
    first_priority['course']
    .map(course_capacity)
    .fillna(30)
)
```

```python

first_priority = first_priority.sort_values(
    ['course', 'percentile'],
    ascending=[True, True],
    kind='stable'
)

first_priority['place'] = (
    first_priority
    .groupby('course')
    .cumcount() + 1
)
```

```python

first_priority['selected'] = (
    first_priority['place'] <= first_priority['capacity']
)

first_wave = first_priority.loc[
    first_priority['selected']
].copy()
```

```python
print('Количество заявок:', len(first_priority))
print('Количество зачислений:', len(first_wave))

display(
    first_wave.sort_values(['course', 'place']).head(20)
)
```

```python
first_wave.groupby('course').size().sort_values(ascending=False)
```

```python
# Лучший перцентиль на каждом курсе
best_percentile = (
    first_wave.groupby('course')['percentile'].transform('min')
)

# Худший перцентиль на каждом курсе
last_percentile = (
    first_wave.groupby('course')['percentile'].transform('max')
)

# Определяем крайние позиции
first_wave['is_best'] = (
    first_wave['percentile'] == best_percentile
)

first_wave['is_last'] = (
    first_wave['percentile'] == last_percentile
)

# Переносим признаки в исходный датафрейм
df['is_first_place'] = df['id'].isin(
    first_wave.loc[first_wave['is_best'], 'id']
).replace({True: True, False: np.nan})

df['is_last_place'] = df['id'].isin(
    first_wave.loc[first_wave['is_last'], 'id']
).replace({True: True, False: np.nan})
```

```python
print(
    'Совпадение лучших:',
    set(df.loc[df['is_first_place'].notna(), 'id'])
    == set(check_df.loc[check_df['is_first_place'].notna(), 'id'])
)

print(
    'Совпадение худших:',
    set(df.loc[df['is_last_place'].notna(), 'id'])
    == set(check_df.loc[check_df['is_last_place'].notna(), 'id'])
)
```
